1. Scenario
2. A retail company has recorded weekly sales (in ₹ thousands) for 4 stores over 12 weeks. The data arrived as a flat 1D array of 48 values with some entries corrupted as np.nan. Your task is to restructure and analyze it.
3. Tasks

* Reshape the 1D array into a (4, 12) matrix — rows = stores, columns = weeks.
* Replace all NaN values with the column (weekly) mean across non-NaN stores.
* Find the top-performing store (highest total annual sales).
* Flag weeks where ALL stores sold below ₹50k (boolean mask).
* Compute cumulative sales per store across the 12 weeks.

```python
import numpy as np

np.random.seed(42)
raw_sales = np.random.randint(30, 91, 48).astype(float)
raw_sales[[5, 14, 23, 33, 40]] = np.nan
```

```python
# 1.
sales = raw_sales.reshape(4, 12)
print("Reshaped matrix:\n", sales)
```

```python

#2.
col_mean = np.nanmean(sales, axis=0)
print('col_mean: ',np.round(col_mean))
cleaned_sales = np.round(np.where(np.isnan(sales), col_mean, sales))
print("cleaned matrix: \n", cleaned_sales)
```

```python
#3.
store_totals = np.round(np.sum(cleaned_sales,axis=1))
top_index = np.argmax(store_totals)   
print("\nStore totals:", store_totals)
print("Top-performing store index:", top_index)
print("Top store total sales:", store_totals[top_index])
```

```python
#4.
print("\nWeeks where all stores sold below ₹50k:")
low_sales_weeks = np.all(cleaned_sales < 50, axis=0)
print(low_sales_weeks)
print('\nverify: \n',np.where(cleaned_sales < 50, 1, 0))
```

```python
#5
cumulative_sales = np.cumsum(cleaned_sales, axis=0)
print(np.round(cumulative_sales))
```
