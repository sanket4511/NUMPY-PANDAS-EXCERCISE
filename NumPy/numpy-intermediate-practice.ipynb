````python
# **Q2.** Create two arrays with the **same values** but different dtypes: `float64` and `float32`.  
# Compare their `.itemsize` and `.nbytes`. How much memory do you save with `float32`?

arr_64 = np.arange(1, 20, 2, dtype=np.float64)
arr_32 = np.arange(1, 20, 2, dtype=np.float32)

print(f"float64 -> Itemsize: {arr_64.itemsize} bytes, Total size: {arr_64.nbytes} bytes")
print(f"float32 -> Itemsize: {arr_32.itemsize} bytes, Total size: {arr_32.nbytes} bytes")

saved_bytes = arr_64.nbytes - arr_32.nbytes
print(f"\nMemory saved with float32: {saved_bytes} bytes")
````

````python
'''**Q3.** Demonstrate the **view vs copy** difference:  
Create `original = np.array([10, 20, 30, 40, 50])`.  
- Make a **view** of the last 3 elements. Change the view's first value to `0`. Print `original`.  
- Make a **copy** of the first 3 elements. Change the copy's first value to `0`. Print `original` again.  
- Explain what you observe.'''

# Q3: View vs copy demo
original = np.array([10, 20, 30, 40, 50])
view = original[2:] 
view[0] = 300
print("\nAfter modifying view:", original)

original = np.array([10, 20, 30, 40, 50])
copy = original[0:3].copy()
copy[0] = 100
print("\nAfter modifying copy:", original)
````

View: 
* A view shares the same memory as the original array.
* If you change the view, the **original array also changes**.

Copy: 
* A copy creates a separate array with its own memory.
* If you change the copy, the **original array does not change**.
