# Series

A Series is a one-dimensional labelled array capable of holding heterogeneous data.

## Components
- Data
- Index
- Data Type

## Facts
- Mutable values
- Immutable size
- Supports vectorized operations

```python
# Create Series from list
s1 = pd.Series([10,20,30], index=['a','b','c'])
print(s1)

# Create Series from dictionary
s2 = pd.Series({'Math':90,'Science':95,'English':88})
print(s2)
```

# DataFrame

A DataFrame is a two-dimensional labelled data structure.

## Properties
1. Heterogeneous Data
2. Size Mutable
3. Labelled Axes
4. Supports Missing Values
5. Powerful Indexing

```python
# List of lists
df1 = pd.DataFrame(
    [['John',25],['Mary',30]],
    columns=['Name','Age']
)
print(df1)

# List of dictionaries
df2 = pd.DataFrame([
    {'Name':'Alice','Age':28},
    {'Name':'Bob','Age':35}
])
print(df2)

# Dictionary of Series
df3 = pd.DataFrame({
    'A': pd.Series([1,2,3]),
    'B': pd.Series([4,5,6])
})
print(df3)
```

# Adding Columns

Methods:
- Assign scalar value
- Assign list
- Assign Series

```python
df = pd.DataFrame({
    'Name':['John','Mary','Alice'],
    'Age':[25,30,28]
})

# Scalar
df['Country'] = 'India'

# List
df['Score'] = [80,90,85]

# Series
bonus = pd.Series([5,10,15])
df['Bonus'] = bonus

print(df)
```

# Rename Columns

rename(columns={}, inplace=False)

Parameters:
- columns
- inplace

```python
# Rename single column
df.rename(columns={'Age':'Employee_Age'}, inplace=True)

# Rename multiple columns
df.rename(columns={
    'Score':'Exam_Score',
    'Bonus':'Extra_Bonus'
}, inplace=True)

print(df)
```

# Indexing and Data Selection

## Types
- loc (label based)
- iloc (position based)

```python
employees = pd.DataFrame({
    'Name':['John','Mary','Alice','Bob'],
    'Department':['IT','HR','IT','Finance'],
    'Salary':[50000,60000,55000,70000]
}, index=['E1','E2','E3','E4'])

print(employees)
```

## loc Examples

```python
# Access row
print(employees.loc['E1'])

# Access column
print(employees.loc[:, 'Name'])

# Multiple rows and columns
print(employees.loc[['E1','E3'], ['Name','Salary']])

# Conditional indexing
print(employees.loc[employees['Salary'] > 55000])
```

## iloc Examples

```python
# Single row
print(employees.iloc[0])

# Single column
print(employees.iloc[:,0])

# Subset
print(employees.iloc[0:3, 0:2])
```

1. Series vs DataFrame
    * Series: A 1D labeled array. It holds a single column of data.
    * DataFrame: A 2D labeled table. It holds multiple rows and columns.
2. loc vs ilocloc: 
    * loc: Label-based selection. You find data using row and column names (e.g., 'E1', 'Age').
    * iloc: Index/Position-based selection. You find data using integer positions (e.g., 0, 1).
3. Why Pandas is Built on NumPySpeed: 
    * NumPy uses fast, pre-compiled C code under the hood.
    * Memory: It stores data in continuous blocks, making it highly efficient.
    * Vectorization: It performs math operations on whole arrays instantly without slow loops.
4. What is Heterogeneous Data?Definition: 
    * Data tables containing mixed data types across different columns.
    * Example: A single DataFrame can store text (Name), integers (Age), and floats (Salary) together.
5. How to Rename Columns
    * Using rename(): Pass a dictionary to change specific names: df.rename(columns={'Old': 'New'}).
    * Using .columns attribute: Overwrite all column names directly: df.columns = ['A', 'B', 'C'].
6. How to Add Columns
    * Scalar Value: Assign one value to all rows: df['New'] = 10.
    * List / Array: Assign position-matched values: df['New'] = [1, 2, 3].
    * Series: Assign index-matched values: df['New'] = pd.Series([1, 2], index=df.index).
7. How Pandas Handles Missing ValuesRepresentation: 
    * Missing values are marked as NaN (Not a Number) or None.
    * Detection: Found using df.isna() or df.isnull().
    * Resolution: Cleaned up using df.fillna(value) to replace them, or df.dropna() to delete them.
8. DataFrame Attributes.
    * shape: Returns a tuple showing total rows and columns (rows, cols)..
    * dtypes: Lists the specific data type of every column..
    * index: Displays the row labels..
    * columns: Displays the column headers.
9. List of Dicts vs Dict of Series
    * List of Dicts: Row-oriented creation. Each dictionary represents a single horizontal row.
    * Dict of Series: Column-oriented creation. Each key is a column header, and the Series provides the vertical column data.
10. When to Choose loc over iloc?
    Choose loc when you want to filter data by explicit names or conditional logical expressions (e.g., df['Age'] > 30). 
    Choose iloc only when you need to slice data by its exact physical layout position (e.g., "get the first 3 rows").
