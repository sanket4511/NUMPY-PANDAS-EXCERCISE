### NP34. Calculate variance and standard deviation.

> 💡 **Hint:** `np.var(arr)` and `np.std(arr)` use `ddof=0`
> (population). Pandas `.var()` and `.std()` use `ddof=1` (sample). Be
> aware of this difference when comparing results.

ddof=0 → divide by N (population) → NumPy default.

ddof=1 → divide by N − 1 (sample, Bessel's correction) → pandas default.

That's why np.var() and Series.var() may return different results for the same data.

```python
print('original data: ')
print(df28)

arr = df28['A'].values
print('\n--- NumPy variance and standard deviation: ---')
print('var: ',np.var(arr,ddof=0))   #Delta Degrees of Freedom.  Population variance/std
print('std: ',np.std(arr,ddof=0))
print('\n--- Pandas variance and standard deviation: ---')
print('var: ',df28['A'].var(ddof=1))    #Sample variance/std
print('std: ',df28['A'].std(ddof=1))
```

A rolling mean (or moving average) is the average of the last w values, where w is the window size.

### NP49. Perform a mini data analysis project.

> 💡 **Hint:** Follow these steps: (1) Generate data with NumPy, (2)
> Store in DataFrame, (3) Clean with Pandas, (4) Engineer features with
> NumPy, (5) GroupBy analysis, (6) Detect outliers, (7) Export summary.

```python
#step 1: Generate data with NumPy---------------------------------------------------------------
np.random.seed(42)

age = np.random.randint(18, 60, 100)
sal = np.random.normal(50000, 12000, 100)
dept = np.random.choice(['HR', 'IT', 'Sales', 'Finance'], 100)
exp = np.random.randint(1, 20, 100)

#step 2: Store in DataFrame---------------------------------------------------------------
df49 = pd.DataFrame({'age': age,'sal': sal,'dept': dept,'exp': exp})

#step 3: Clean with Pandas----------------------------------------------------------------
print('check null values: \n',df49.isnull().sum())

df49 = df49.drop_duplicates()

print('\n--------- Cleaned data: ------------')
print(df49.head())

#step 4: Engineer features with NumPy------------------------------------------------------------
# Salary category
df49['salary_level'] = np.where(
    df49['sal'] >= 50000,
    'High',
    'Low'
)

# Experience category
condition = [df49['exp'] < 5, df49['exp'] < 10, df49['exp'] >= 10]
df49['exp_level'] = np.select(condition,['Beginner', 'Intermediate', 'Expert'], default='unknown')
print('\n----------------- After feature engineering: -------------------')
print(df49.head())

#step 5: GroupBy analysis--------------------------------------------------------------------------
# Average salary by department
avg_sal_by_dept = df49.groupby('dept')['sal'].mean()
print('\n----- GroupBy analysis ------')
print('Average salary by department: \n',avg_sal_by_dept)

# Average experience by departmen
avg_exp_by_dept = df49.groupby('dept')['exp'].mean()
print('\nAverage experience by department: \n',avg_exp_by_dept)

# Multiple aggregations
summary = df49.groupby('dept').agg({
    'sal': ['mean', 'min', 'max'],
    'exp': ['mean', 'sum']
})

print('\n---------------------Aggregate Summary: ---------------------------')
print(summary)

#step 6: Detect outliers-------------------------------------------------------------------
arr49 = df49['sal'].values

q1 = np.percentile(arr49, 25)
q3 = np.percentile(arr49, 75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = df49[(arr49 < lower_bound) | (arr49 > upper_bound)]

print('\nOutliers:')
print(outliers)

#step 7: Export summary----------------------------------------------------------------------

summary.to_csv('department_summary.csv')
print('\nFile saved as department_summary.csv')
```

### NP50. Build an end-to-end pipeline using NumPy + Pandas.

> 💡 **Hint:** Chain all steps together:  
> 1. `np.random.seed(X)` → generate raw data  
> 2. `np.clip` → clean invalid values  
> 3. `np.log`, `np.square`, `np.where` → feature engineering  
> 4. `df.groupby().agg(...)` → statistical analysis  
> 5. IQR + `np.clip` → outlier removal  
> 6. `df[features].to_numpy(dtype=np.float32)` → ML feature matrix  
> 7. Export with `df.to_csv(...)`

```python
# Step 1: generate raw data-------------------------------------------------------------
np.random.seed(42)
data = {
    "age": np.random.randint(15, 90, size=100),         
    "salary": np.random.uniform(20000, 150000, size=100),
    "dept": np.random.choice(["Sales", "IT", "HR"], size=100)
}
df = pd.DataFrame(data)

# Step 2: clean invalid values-----------------------------------------------------------
df["age"] = np.clip(df["age"], 18, 65)

# Step 3: Feature engineering using np.log, np.square, and np.where
df["log_salary"] = np.log(df["salary"])
df["age_squared"] = np.square(df["age"])
df["is_senior"] = np.where((df["age"] > 50) & (df["salary"] > 80000), 1, 0)

# Step 4: Statistical analysis using groupby and agg---------------------------------------
dept_summary = df.groupby("dept").agg(
    avg_salary=("salary", "mean"),
    max_age=("age", "max"),
    total_seniors=("is_senior", "sum")
)
print("--- Step 4: Department Summary ---")
print(dept_summary, "\n")

# Step 5: IQR + np.clip ------------------------------------------------------------------------
q1, q3 = np.percentile(df["salary"], [25, 75])
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
df["salary"] = np.clip(df["salary"], lower_bound, upper_bound)

# Step 6: extract feature matrix (float32 datatype)---------------------------------------------
features = ["age", "salary", "log_salary", "age_squared", "is_senior"]
X = df[features].to_numpy(dtype=np.float32)

print("--- Step 6: ML Feature Matrix Shape ---")
print(f"Matrix shape: {X.shape} | Data type: {X.dtype}\n")

# Step 7: Export final processed DataFrame to CSV-----------------------------------------------
df.to_csv("processed_pipeline_output.csv", index=False)
print("Processed pipeline exported successfully to CSV...")
```
